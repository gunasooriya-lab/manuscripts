```python
from ase.db import connect
from ase.io import read
import pandas as pd
```

With this manuscript, we have provided 9 databases(DB) for strain effects on IrO2(110), IrO2(100) and IrO2(101) surfaces considered in the study. For each surface, strain effects under low coverage conditions, in the presence of a nearby Ir vacancy and under high coverage conditions were evaluated.

Each database contains following data:

| Name | Descriptions |
| --- | --- | 
| atoms | ASE atoms object | 
| configuration |  surface configuration ie. 100_x_y, 100_x_y_Ir_vac| 
| strain | strain applied on the surface ie.3%_surface, 3-%_surface| 
| calculation | OER intermediate calculation ie. O*, OH* etc. | 
| composition | composition of atoms | 
| num_sites | Number of Atoms | 
| final_energy | Final electronic energy | 




# How to connect to the database files

# Extracting specific data

```python
# Extracting specific data ex: strain='3%_surface', calculation = 'O' of the considered IrO2(110) low coverage
for row in db.select(strain='3%_surface', calculation = 'O'):      
    atoms = row.toatoms()                          
    print('Method 1: ', row.strain, row.calculation)                        
    print('Method 2: ', row['strain'], row['calculation'])  
    print('Method 3: ', row.get('strain'), row.get('calculation'))

    data = row.data
    print(data) # printing additional data such as composition, number of sites, and final energy
    print('final_energy:', data['final_energy'])
```
