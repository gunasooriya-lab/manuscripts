```python
from ase.db import connect
from ase.io import read
import pandas as pd
```

With this manuscript, we have provided 10 databases(DB) for each SAC configuration considered in the study. These SAC configurations include surface, subsurface, bulk and adatom configurations on SnO2(110) and SnO2(100) surfaces. 

Each database contains following data:

| Name | Descriptions |
| --- | --- | 
| atoms | ASE atoms object | 
| configuration |  SAC configuration | 
| SAC | TM SAC ie. Ir, Pd| 
| calculation | OER intermediate calculation ie. O*, OH* etc. | 
| composition | composition of atoms | 
| num_sites | Number of Atoms | 
| final_energy | Final electronic energy | 




# How to connect to the database files

```python
# Connect to the database using the db file
db = connect('SnO2_100-adatom.db')

# Check to see if properly working by printing each row of data
for row in db.select():
    print(row)
```

# How to visualize the avaialble key-value pairs

```python
# Connect to the database using the db file
db = connect('SnO2_100-adatom.db')

# Viewing a specific row
entry = db.get(1)
print(entry)
```

# Extracting specific data

```python
# Extracting specific data ex: SAC='Ir', calculation = 'O' of the considered configuration
for row in db.select(SAC='Ir', calculation = 'O'):      # selecting row(s) SAC='Ir', calculation = 'O'   
    atoms = row.toatoms()                          
    print('Method 1: ', row.SAC, row.calculation)                        
    print('Method 2: ', row['SAC'], row['calculation'])  
    print('Method 3: ', row.get('SAC'), row.get('calculation'))

    data = row.data
    print(data) # printing additional data such as composition, number of sites, and final energy
    print('final_energy:', data['final_energy'])
```
